```python
'''Problem Statement: Create a Python program that implements a simple number 
guessing game using a while loop. The program should make use of control 
statements such as else, break, and continue.'''
```

```python

import random
secret_number=random.randint(1,10)
print('Welcome to the Guessing Game!')
print('Can you guess the secret number')
attempts=3
while attempts>0:
    guess=int(input(f"\nYou have {attempts} attempts left.Enter your guess: "))
    if not(1<=guess<=10):
        print('Your guess is out of range.Please guess a number between 1 and 10.')
        continue
    attempts-=1
    if guess==secret_number:
        print(f"Congratulations! You guessed the correct number:{secret_number}")
        break
    elif guess>secret_number:
        print("Too high.Try again")
    else:
        print("Too low.Try again")
else:
    print(f"\nGame Over! The secret number was {secret_number}.")
    print("Better luck next time!")
        
           
```

```python
'''Create a Python program that generates and prints a 
multiplication table (from 1 to 10) for a given number using a for loop and the range 
function. '''
```

```python
number = int(input("Enter the number for which you want the multiplication table:  "))
for i in range(1, 11):
     result = number * i
     print(f"{number} x {i} = {result}")
```

```python
'''Create a Python program that calculates the Body Mass Index 
(BMI). '''
```

```python
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi
user_weight = float(input("Enter your weight in kg: "))
user_height = float(input("Enter your height in m: "))
calculated_bmi = calculate_bmi(user_weight, user_height)
print(f"\nYour BMI is: {calculated_bmi:.2f}")
```
